```python
filepath = Path("data/REV1-SCOPUS-1/2026-03-26_Export.ris")
with open(filepath, "r", encoding="utf-8") as f:
    entries = rispy.load(f)
```

```python
# Turn into dataframe
df = pd.DataFrame(entries)
print(f"Columns: {df.columns}")
df.head()
```

```python
# Show where custom7 is nan
df.loc[df["custom7"].isna(), "custom2"]
```

```python
# Filter by year > 2021, type is Jurnal and has DOI
filtered_df = df[(df["year"].astype(int) > 2021) & (df["type_of_reference"] == "JOUR") & (df["doi"].notna())]
print(f"Filtered entries: {len(filtered_df)}")
filtered_df.head()
```
